3. Code the Algorithm: Grade Calculator

```python
def calculate_grade(score1, score2, score3):
    """
    Calculate letter grade from three test scores
    
    Args:
        score1, score2, score3: Test scores (integers)
    
    Returns:
        grade: Letter grade (string)
    """
    average_score = (score1 + score2 + score3) / 3
    
    if average_score >= 90:
        grade = 'A'
    elif average_score >= 80:
        grade = 'B'
    elif average_score >= 70:
        grade = 'C'
    elif average_score >= 60:
        grade = 'D'
    else:
        grade = 'F'

    return grade

print("Test 1:", calculate_grade(95, 92, 88))  # Should be 'A'
print("Test 2:", calculate_grade(85, 80, 82))  # Should be 'B'
print("Test 3:", calculate_grade(75, 70, 72))  # Should be 'C'
print("Test 4:", calculate_grade(65, 60, 62))  # Should be 'D'
print("Test 5:", calculate_grade(55, 50, 52))  # Should be 'F'
```

## 1.2 Variables and Data Types

<strong>Variable: </strong> named storage location in memory for data. The more specific the name, the better

* naming conventions... lowercase first letter then subsequent words are capitalized (e.g. favoriteFood)

<strong>Declaration: </strong> creating a variable with a type.

<strong>Initalization: </strong> giving the variable a value at creation. (ex. initial_price = 5)

<strong>Assignment: </strong> giving a value later, after declaration

<em> The difference between these three is that declaration creates the variable (int number), assignment stores the value in the variable (number = 10), while intialization combines the two (int age = 16). </em>

<strong>Primitive Type: </strong> basic data types built into Java (int, double, boolean, char), they store the actual value and cannot be `null`

<strong>Reference Type: </strong> points to objects; more complex (Strong, Scanner). they can be `null`. 

<em> Primitive types don't have methods while Reference types are objects and will have methods associated with them </em>

<strong>Final Variables: </strong> These will <em> never </em> change. just put final at the beginning and give it a value. `final int MAX_STUDENTS = 30;` USE ALL CAPS (e.g. BIRTH_YEAR)

#### Java Data Types (AP CSA Focus)

| Type      | Meaning | Example | Notes |
|-----------|---------|---------|-------|
| `int`     | Whole numbers | `int year = 2025;` | Range: about -2 billion → 2 billion |
| `double`  | Decimal numbers | `double price = 4.99;` | More precise than `float` |
| `boolean` | True/False | `boolean done = false;` | Used for conditions |
| `char`    | Single character | `char grade = 'A';` | Always single quotes |
| `String`  | Text (sequence of chars) | `String name = "Paul";` | Not primitive (it’s a class) |

#### Choosing the correct Data Type:
> For numbers:
* Whole Numbers (1, 2, 100) -> use `int`
* Decimal Numbers (1.5, 3.14) -> use `double`

> For words and letters:
* One Letter ('A' , 'B') -> use `char` SINGLE QUOTES
* Words or sentences -> use `String` DOUBLE QUOTES 

> For yes/no questions:
* True or False -> use `boolean`
