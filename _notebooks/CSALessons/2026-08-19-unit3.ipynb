## Unit 3.3 — Anatomy of a Class

#### 1. Console Based Menu
- **Imports used:**
  - `java.util.Scanner` → reads user input from the terminal.
  - `java.lang.Math` → generates random numbers (`Math.random()`).
  - `System` class → static methods `System.out.print` / `System.out.println` for output.
- **2D array** `COLORS` stores pairs of `{name, ANSI code}` used to color terminal text.
- **Control structure:** `switch` is used in the `action()` method to decide what to do based on the user's menu choice.
- **Constructor** `Menu()` runs automatically when `new Menu()` is called. It:
  1. Creates a `Scanner` object.
  2. Prints the menu.
  3. Loops, reading input, until the user quits.
- **Key idea:** instantiating `new Menu()` hands control of the "frontend" (console interaction) over to that object — similar in concept to how a web framework (e.g. Spring Boot) hands control to a controller/route.

#### 2. Desktop GUI Menu
- Uses `javax.swing.*`, `java.awt.*`, `java.awt.event.*`, `java.util.Timer`, `java.util.TimerTask`.
- **1D array** `MENUS` stores menu option names.
- **Control structure:** `if / else if / else` (instead of `switch`) decides the action.
- Uses a `Timer` + `TimerTask` so the "loading bar" animation can run repeatedly **without freezing** the main GUI thread.
- `MenuJFrame extends JFrame implements ActionListener` — the class inherits from `JFrame` and implements the `ActionListener` interface so it can respond to menu clicks via `actionPerformed(ActionEvent e)`.

##### Console vs. GUI — Key Differences
| | Console Menu | GUI Menu |
|---|---|---|
| Input | `Scanner` reads typed input | Mouse clicks fire `ActionEvent`s |
| Output | Printed text (with ANSI color codes) | Rendered widgets (`JLabel`, `JFrame`, HTML-formatted text) |
| Control structure | `switch` | `if / else if / else` |
| Array dimension | 2D (`COLORS`) | 1D (`MENUS`) |
| Blocking behavior | Loops and blocks on `sc.nextInt()` until input arrives | Event-driven; `Timer`/`TimerTask` allows non-blocking repeated actions |

#### 3. Object Superclass
- **Every** class that does **not** use `extends` implicitly extends `Object`.
- This means every class automatically inherits methods like:
  1. `getClass()`
  2. `toString()`
  3. `equals()`
- If you want to redefine (override) one of these methods, it **must** be declared `public` (matching `Object`'s access level) and marked `@Override`.
  - `getClass()` **cannot** be overridden.
- Example of what goes wrong without `@Override`/`public`:
  ```java
  // this will return an error
  class Shape {
      String toString(){       // wrong: not public, missing @Override
          return "Shape";
      }
  }
  ```
  ```java
  // this is fine
  class Shape {
      @Override
      public String toString(){
          return "Shape";
      }
  }
  ```

#### 4. Introduction to Inheritance
- **Inheritance**: one class (subclass) reuses fields/methods from another (superclass), via the `extends` keyword.
- **Superclass (parent)**: provides common attributes/behavior.
- **Subclass (child)**: inherits from the parent, and can add or modify behavior.
- Syntax:
  ```java
  public class Subclass extends Superclass {
      // subclass-specific fields/methods
  }
  ```
- Example: `Vehicle` is the superclass with `miles`, `milesTraveled()`, and `hasMovement`. Both `Car` and `Plane` `extend Vehicle`, so they automatically get `hasMovement` and `milesTraveled()` without rewriting them.

```java
public class PopcornHack_ObjectMethods {

    // Shape does NOT use "extends", so it implicitly extends Object.
    static class Shape {

        // We do NOT touch getClass() -> it stays exactly as Object defines it.
        // (getClass() actually can't be overridden anyway.)

        // We DO override toString(), which Object also provides by default
        // (the default would print something like "Shape@1b6d3586").
        @Override
        public String toString() {
            return "Shape (a custom toString from the Shape class)";
        }
    }

    public static void main(String[] args) {
        Shape s = new Shape();

        // 1. UNCHANGED Object method: getClass()
        //    This is the exact getClass() inherited from Object, untouched.
        System.out.println("Unchanged Object method -> getClass(): " + s.getClass());

        // 2. CHANGED Object method: toString()
        //    This is our own version, overriding Object's default toString().
        System.out.println("Overridden Object method -> toString(): " + s.toString());

        // Bonus: show what equals() (also inherited, unchanged) does by default -
        // reference comparison, since we didn't override it either.
        Shape s2 = new Shape();
        System.out.println("Unchanged Object method -> equals(): " + s.equals(s2)
                + " (false, because equals() wasn't overridden, so it compares references)");
    }
}
```

Every class inhereits methods from its <strong>parent class</strong>. These methods can then be called in the child class and can also be <strong> overridden</strong>. 

#### Shape Class
```java
public class Shape {
    protected String name;
    private int length;
    private int width;

    // Default constructor
    public Shape() {
        this.name = "shape";
        this.length = 10;
        this.width = 5;
    }

    // Parameterized constructor
    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    // Getter methods
    public String get_name() {
        return this.name;
    }

    public int get_length() {
        return this.length;
    }

    public int get_width() {
        return this.width;
    }

    // Setter methods
    public void set_name(String n) {
        this.name = n;
    }

    public void set_length(int a) {
        this.length = a;
    }

    public void set_width(int b) {
        this.width = b;
    }

    // Method to calculate the area
    public double calc_area() {
        return this.length * this.width;
    }

    // Method to calculate the perimeter
    public double calc_perimeter(){
        return 2*this.length + 2*this.width;
    }

    // Method to print the shape
    public void print_shape() {
        System.out.println("Shape: " + this.name);
    }

    // Additional method to print something
    public void print_something() {
        System.out.println("This is a shape");
    }
}

public class Driver {
    public static void main(String[] args) {
        Shape s1 = new Shape();
    }
}
```

#### The Super Keyword

The `super` keyword calls a **parent class method** from a child class.

Using super to Call Parent Methods

```java
public class Triangle extends Shape {
    private int side1, side2, side3;

    @Override
    public void print_shape() {
        super.print_something();  // Call parent's print_something()
        print_something();        // Call this class's print_something()
        System.out.println("Shape: " + this.name);
    }

    @Override
    public void print_something() {
        System.out.println("This is a triangle");
    }
}

// Test
Triangle t2 = new Triangle("triangle", 3, 4, 5);
t2.print_shape();

// Output:
// This is a shape        (from super.print_something())
// This is a triangle     (from this.print_something())
// Shape: triangle        (from print_shape())
```

**Order of execution:** Parent method → child method → rest of code

Using super in Constructors

```java
public class Triangle extends Shape {
    private int side1, side2, side3;

    public Triangle(String name, int s1, int s2, int s3) {
        super(name, 0, 0);  // Calls parent constructor to set name
        this.side1 = s1;
        this.side2 = s2;
        this.side3 = s3;
    }
}
```

**Important:** `super()` must be the **first line** in the constructor.



#### Important Rules

##### Overriding Rules

✅ **Can override** a parent method to provide child-specific behavior  
✅ **Must use same signature** (method name, parameters, return type)  
✅ **Can increase access** (private → public), but not restrict (public → private)  
✅ **Use @Override** annotation for verification  
✅ **final methods** cannot be overridden

##### Super Usage Rules

✅ **super.method()** calls parent method  
✅ **super()** calls parent constructor (must be first line)  
✅ Use when you want to combine parent + child behavior  


Common Pitfall

❌ **DO NOT** call method without `super` in override:

```java
// WRONG: Infinite loop
public void print_shape() {
    print_shape();  // Calls itself infinitely!
}

// CORRECT: Use super to call parent
public void print_shape() {
    super.print_shape();  // Calls parent
}
```


Complete Triangle Example

```java
public class Triangle extends Shape {
    private int side1;
    private int side2;
    private int side3;

    public Triangle() {
        this.name = "triangle";
        this.side1 = 1;
        this.side2 = 2;
        this.side3 = 3;
    }

    public Triangle(String name, int s1, int s2, int s3) {
        super(name, 0, 0);
        this.side1 = s1;
        this.side2 = s2;
        this.side3 = s3;
    }

    @Override
    public double calc_area() {
        double s = (side1 + side2 + side3) / 2.0;
        return Math.sqrt(s * (s - side1) * (s - side2) * (s - side3));
    }

    @Override
    public double calc_perimeter() {
        return side1 + side2 + side3;
    }

    @Override
    public void print_shape() {
        super.print_something();
        print_something();
        System.out.println("Shape: " + this.name);
    }

    @Override
    public void print_something() {
        System.out.println("This is a triangle");
    }
}

// Test
Triangle t = new Triangle("triangle", 3, 4, 5);
t.print_shape();
System.out.println("Area: " + t.calc_area());
System.out.println("Perimeter: " + t.calc_perimeter());
```
