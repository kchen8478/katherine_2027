## Corrections

**Q15 - substrings in a while loop.** `str = "lookout"`, `target = "o"`. Tracing each pass:

| pass | `j` | `str.substring(j)` printed | `str` after `substring(1)` | next `j` |
|---|---|---|---|---|
| 1 | 1 | `ookout` | `okout` | 0 |
| 2 | 0 | `okout` | `kout` | 1 |
| 3 | 1 | `out` | `ut` | -1 (stop) |

Output: `ookout okout out` (A). I dropped pass 2 because I assumed the next `"o"` could not be at index 0. The loop condition is `j >= 0`, so index 0 counts.

**Q21 - missing inner loop header.** Row `j` must print `j` exactly `j + 1` times (row 0 prints one `0`). `for (int k = 0; k <= j; k++)` runs `j + 1` times (B). My answer, `k < j`, runs only `j` times, so the first row is empty and the rest are one short. Rule I am keeping: **count iterations as `end - start` for `<`, and `end - start + 1` for `<=`.**

**Q22 - Pet class does not compile.** `getInfo()` returns `pName + ", " + pSpecies`, but `pName` and `pSpecies` are constructor **parameters**: their scope ends when the constructor ends (D). It should use the instance variables `name` and `species`. `private` instance variables are not an error; they are good practice.

**Q25 - static count in Shoe.** `count` is `static`, so all `Shoe` objects share one counter, but only the **two-parameter** constructor runs `count++`. `s1` and `s3` use it and `s2` uses the no-argument constructor, so `getCount()` returns `2` (C), not 3.

**Q28 - count equal neighbors.** The loop compares each character with the next one in `"abbcdddeff"`: `bb`, `dd`, `dd`, `ff` gives **4** (C). I counted "groups of repeated letters" (`bb`, `ddd`, `ff`) and then miscounted; `ddd` contains **two** adjacent equal pairs.

**Q29 - recursive printNums(10).** Each call recurses **before** it prints, so nothing prints until the chain stops:

```text
printNums(10) -> printNums(-20) -> printNums(40) -> printNums(-80) -> printNums(160) stops (160 >= 50)
prints on the way back:  -80  40  -20  10
```

Answer A. My answer (B) is what happens if the print comes before the recursive call.

**Q30 - minOfThree.** Choice A returns `b` whenever `a > b` without checking whether `c` is even smaller (for `30, 20, 10` it returns 20). Choice D checks that the returned value is `<=` both others before returning it, falling through to `c` otherwise.
