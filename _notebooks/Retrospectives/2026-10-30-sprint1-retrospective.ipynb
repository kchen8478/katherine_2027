## What I Know (and Can Explain Myself)

**The POJO -> JPA -> Repository -> REST flow.** My peer assessment feature follows the same path as the Jokes example:

```text
PeerAssessment (@Entity) -> PeerAssessmentRepository (JpaRepository) -> PeerAssessmentController (/api/peer-assessments) -> JSON
```

- `PeerAssessment` is a POJO mapped to the `peer_assessment` table with `@Entity`, `@Table`, and `@Column`. It stores who assessed whom, for which assignment and submission, a 1-5 rating, written feedback (`columnDefinition = "TEXT"`), and a timestamp.
- `PeerAssessmentRepository` extends `JpaRepository` and uses derived queries such as `findBySubmissionId` and `findByAssignmentIdAndAssessedPersonId`. Spring Data writes the SQL from the method name.
- `PeerAssessmentController` exposes `POST /api/peer-assessments` and GET endpoints by submission, assessor, assignment, and assignment + person.
- I also extended an existing model: `Assignment` gained an `assignmentType` field (`all_assignments` or `sprints`), a new constructor, a `GET /api/assignments/type/{type}` endpoint, and seed data in `ModelInit`. Changing a shared model, not only adding my own, taught me that other code (seed data, controllers) depends on its constructors.

**AP CSA connection.** I wrote `PeerAssessment`'s getters and setters by hand, which is exactly the AP "accessor and mutator" pattern; Lombok's `@Data` would generate the same methods.

**AP Unit 1 foundations.** My Unit 1 notes cover 1.1-1.15, including choosing data types, arithmetic and assignment, and the `Scanner` `nextInt()` / `nextLine()` pitfall, which I can explain and demonstrate.
