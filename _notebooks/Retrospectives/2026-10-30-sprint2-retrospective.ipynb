## What I Know

- **What makes a lesson usable, not just correct.** Restructuring 15 lessons showed me that consistency matters as much as content: every lesson needs the same structure, a clear description, and an order that builds from the previous one. Learners lose time when lessons are out of order or inconsistently formatted.
- **Practice has to be runnable.** Adding CodeRunner exercises to lessons 1.5-1.8 turned reading into doing; students can run and change code inside the lesson. Assignment tags connect each exercise to the course's assignment tracking.
- **The Unit 1 concepts I worked on.** Revising 1.5 (casting and ranges) and 1.6 (compound assignment operators) made me explain integer division, casting, and operator shortcuts precisely enough for someone else to learn from.
- **The system topics I picked.** I chose topics that are part of the OCS system itself: how login uses a JWT stored in an httpOnly cookie, how game and AI state is persisted, and how utility classes (like Tailwind) are generated with SASS.

## What Improved

- **From consuming lessons to maintaining them.** In Sprint 1 I took notes on lessons; in Sprint 2 I improved the shared lessons that the whole class uses.
- **Version control as a team practice.** My lesson changes went through small, described commits in the shared OCS Pages repo over two weeks instead of one large drop.
- **Connecting lessons to the system.** My topic choices tie AP concepts to real OCS code. The JWT lesson is planned around selection and iteration in the login flow, starting from the real `Set-Cookie` response the server sends.

## What Still Needs Work

- **Communication.** Most of my Sprint 2 work happened quietly in commits. I did not write down the POV statement, the How Might We question, or the feedback I received and the change it caused, so the people I worked with and my teacher cannot see my thinking or the LxD "Test" step. Next sprint I will share what I am working on and where I am stuck early, ask questions in the open instead of waiting, and record each learner question and the change I made because of it.
- **Tangible tech work and coding.** My Sprint 2 contributions were mostly formatting, descriptions, lesson order, and adding practice cells, not code I wrote myself. My system-topic lessons are still outlines: the JWT/cookies page has a `Set-Cookie` example and an empty homework hack heading, and the AI pipeline and SASS utilities pages have only titles. I need working code I can point to, such as annotated Java and Spring examples for the JWT login flow that run, plus the popcorn hack, homework hack, and grading plan the lesson checklist asks for.
